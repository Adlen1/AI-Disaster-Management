# Prevention Phase — Data Ingestion & Curation (Abid Extended Dataset)

## Objective

The purpose of this notebook is to curate the Extended Algerian Forest Fires Dataset before it is used for feature engineering and machine learning.

The goal is to ensure that the dataset is clean, consistent, and reproducible while preserving all relevant information.

---

## Input

Dataset:

- Extended Algerian Forest Fires Dataset
- Source: IEEE DataPort
- 1215 observations
- Nine Algerian regions
- Observation years: 2012, 2017, and 2021

Location:

```
data/Prevention/raw/abid_extended_forest_fires.xlsx
```

---

## Curation Steps

### 1. Initial inspection

The dataset structure was inspected by checking:

- dataset dimensions
- column names
- data types
- first and last rows

---

### 2. Date processing

The `DATE` column was converted to the datetime format.

Day, month, and year were extracted for temporal analysis.

---

### 3. Region standardization

The original column named `Bejaia 2021` was renamed to `Region`.

The embedded year was removed because it is already represented by the `DATE` column.

Examples:

| Original | Curated |
|-----------|---------|
| Bejaia 2021 | Bejaia |
| Tipaza2017 | Tipaza |
| Tizi-Ouzou2021 | Tizi-Ouzou |

---

### 4. Fire label standardization

The `Classes` column was normalized by:

- converting text to lowercase
- removing surrounding whitespace
- creating a binary label

Mapping:

- fire → 1
- not fire → 0

---

### 5. Data validation

The following validations were performed:

- missing values
- duplicate records
- invalid observation dates
- weather variable ranges
- fire class distribution

---

### 6. Exploratory Data Analysis

EDA included:

- descriptive statistics
- fire class balance
- fire occurrence by month
- regional distribution
- yearly distribution
- weather variable summary

---

## Data Quality Issues

During validation, six observations were found with inconsistent dates.

The years stored in the `DATE` column did not match the dataset description or the region metadata.

These records were flagged for review before creating the final curated dataset.

---

## Output

Curated dataset:

```
data/prevention/processed/abid_extended_clean.csv
```

This dataset is the input for the next stage:

- Feature Engineering
- Fire Prediction Modeling

---

# Clean the column names

````python
print(df.columns.tolist())

df.columns = df.columns.str.strip()

print(df.columns.tolist())
````

````python
df["DATE"] = pd.to_datetime(df["DATE"])
df.info()
````

# Inspect the target

````python
df["Classes"].unique()

df["Classes"] = (
    df["Classes"]
      .astype(str)
      .str.strip()      # remove leading/trailing spaces
      .str.lower()      # lowercase everything
)

df["Classes"].unique()

df["fire_label"] = df["Classes"].map({
    "not fire": 0,
    "fire": 1
})

df[["Classes", "fire_label"]].head()
````

# Data validation:
 According to the dataset paper, observations exist only for
 2012, 2017 and 2021.

 Six records contained invalid years (2025 and 1970 date) in lines :
 721 / 723 / 724 / 725 / 726 / 818 
 After manual verification against the original Excel file,
 These records are dropped

````python
type(df.loc[721, "DATE"])
````

````python
df[df["DATE"].dt.year == 2025]
````

````python
df[df["DATE"].dt.year == 1970]
````

````python
valid_years = [2012, 2017, 2021]

invalid_dates = df[~df["DATE"].dt.year.isin(valid_years)]

print(invalid_dates)

df = df[df["DATE"].dt.year.isin(valid_years)]
````

````python
df = df.rename(columns={"Bejaia 2021": "Region"})
````

````python
df["Region"] = (
    df["Region"]
      .astype(str)
      .str.strip()                    # Remove leading/trailing spaces
      .str.replace("_", " ", regex=False)
      .str.replace(r"\d{4}", "", regex=True)   # Remove 4-digit years
      .str.strip()
)
````

````python
weather_cols = ["Temperature", "RH", "Ws", "Rain", 
                "FFMC", "DMC", "DC", "ISI", "BUI", "FWI"]
print(df[weather_cols].describe().round(2))
````
